# Semana del 12 de noviembre · el explorador de Plan2Explore junta los datos

**Lo que pidió el profe:** repetir el experimento random, pero que los datos los junte un agente que **explora** en vez de uno que elige al azar. El explorador es Plan2Explore, traído del repo de Pedro (`pedropalmav/her-dream`).

**Cómo explora.** Ocho modelos chicos intentan predecir el paso siguiente del world model. Donde no se ponen de acuerdo, el modelo todavía no entiende qué pasa, y ahí el explorador recibe recompensa. Así va a buscar lo que el modelo aún no conoce. Las galletas no le dan recompensa.

**Qué aprende.** Igual que en los runs random: con los datos del explorador se entrenan el world model, el actor y el crítico (`model.train_actor_critic=True`). En las evaluaciones y en el video juega el **actor aprendido**, nunca el explorador.

**Comparable con el random de la semana del 22.** Misma configuración: mapa de pasillos cortos, vista de 5×5, semilla 0, decoder Dreamer, imagen 64×64, `TRAIN_RATIO = 128`, buffer de 500.000 en RAM, muestreo uniforme del buffer, entropía original y evaluación cada 25.000 pasos con 5 partidas. Solo cambia quién junta los datos: `model.plan2explore=True` en vez de `trainer.random_policy=True`. Con `VISTA` (sección 6) se puede repetir con otra vista.

**Carpeta nueva.** Este run guarda en `galleta/semana12/`, con sufijo `_explorador`; empieza desde cero la primera vez y retoma su propio checkpoint al repetirlo. Los runs anteriores no se tocan.

**Comprobación antes de entrenar.** La sección 6 verifica con Hydra que junta datos el explorador y que el random está apagado, y rechaza una carpeta con otro tipo de run.

**Qué resultado mirar.** `episode/eval_score` mide las galletas del actor aprendido: es la medida principal. `episode/score` mide las galletas que vio el explorador. La sección 11 compara ambas con el random; las del explorador se notan mucho antes que el aprendizaje del actor.

## 6. Configurar el run y comprobar que junta datos el explorador

La recolección está fijada en `model.plan2explore=True`, con `trainer.random_policy=False`. Mantén la misma configuración al retomar: cambiar la vista exige otra carpeta, y el nombre de la carpeta ya incluye la vista. La verificación comprueba también que un checkpoint anterior en la carpeta sea de un run con explorador.

`PASOS` permite ampliar el objetivo y `PARAR_EN_PASO` permite detenerse guardando todo. El modelo se guarda cada 10.000 pasos y el buffer cada 100.000. `APAGAR_AL_TERMINAR = True` libera la sesión al terminar correctamente.

```python
from pathlib import Path
import json
import shlex
import subprocess

CASO = "explorador"
VISTA = 5  # None: mapa completo; 3, 5 o 7: ventana local
TAREA = "cookie_full_hall5" if VISTA is None else f"cookie_partial_hall5_view{VISTA}"
PASOS = 500_000
TRAIN_RATIO = 128
IMAGEN = 64
REP_LOSS = "dreamer"
SEMILLA = 0
BUFFER = 500_000
REWARD_SLICES = 0  # muestreo uniforme del buffer, igual que en los runs random
ACT_ENTROPY = 3e-4  # valor original, para el actor y para el explorador
GUARDAR_BUFFER_CADA = 100_000
EVALUAR_CADA = 25_000
PARTIDAS_EVAL = 5
PARAR_EN_PASO = None
APAGAR_AL_TERMINAR = True
REPO = "/content/r2dreamer-cookie"

assert CASO == "explorador"
LOGDIR = f"/content/drive/MyDrive/galleta/semana12/{TAREA}_s{SEMILLA}_explorador"
# El run random con la misma vista, para la comparación de la sección 11.
LOGDIR_RANDOM = f"/content/drive/MyDrive/galleta/semana22/{TAREA}_s{SEMILLA}_random"

OPCIONES = (
    f"env=cookie env.task={TAREA} env.train_ratio={TRAIN_RATIO} 'env.size=[{IMAGEN},{IMAGEN}]' "
    f"seed={SEMILLA} model.rep_loss={REP_LOSS} "
    "model.plan2explore=True trainer.random_policy=False model.train_actor_critic=True "
    f"buffer.reward_slices={REWARD_SLICES} model.act_entropy={ACT_ENTROPY} "
    f"trainer.video_pred_log=True buffer.max_size={BUFFER} buffer.storage_device=cpu"
)


def verificar_explorador(argumentos):
    # Se comprueba la configuración efectiva que Hydra entrega al entrenador.
    # Un False, incluso añadido al final del comando, impide iniciar el run.
    comprobacion = r"""
import json
import sys
from pathlib import Path
from hydra import compose, initialize
from omegaconf import OmegaConf

with initialize(version_base=None, config_path="configs"):
    cfg = compose(config_name="configs", overrides=json.loads(sys.argv[1]))
    OmegaConf.resolve(cfg)
assert cfg.model.plan2explore is True, "ERROR: model.plan2explore debe ser True"
assert cfg.trainer.random_policy is False, "ERROR: trainer.random_policy debe ser False"
assert cfg.model.train_actor_critic is True, "ERROR: el actor y el crítico deben seguir aprendiendo"
assert cfg.env.task == sys.argv[2], "ERROR: tarea distinta de la seleccionada"
assert cfg.buffer.reward_slices == 0, "ERROR: esta comparación usa muestreo uniforme del buffer"
assert cfg.model.act_entropy == 3e-4, "ERROR: esta comparación conserva la entropía original"
carpeta = Path(cfg.logdir)
anterior = carpeta / ".hydra/config.yaml"
if anterior.exists():
    guardado = OmegaConf.load(anterior)
    assert guardado.model.get("plan2explore") is True, "ERROR: esta carpeta contiene un run sin explorador"
    assert guardado.trainer.random_policy is False, "ERROR: esta carpeta contiene un run random"
    assert guardado.env.task == cfg.env.task, "ERROR: esta carpeta pertenece a otra tarea"
    assert guardado.buffer.reward_slices == cfg.buffer.reward_slices, "ERROR: cambió el muestreo del buffer"
    assert guardado.model.act_entropy == cfg.model.act_entropy, "ERROR: cambió el coeficiente de entropía"
elif (carpeta / "latest.pt").exists():
    raise RuntimeError("Hay un checkpoint sin configuración verificable; usa otra carpeta")
print("VERIFICADO: model.plan2explore=True y trainer.random_policy=False")
print("RECOLECCIÓN: el explorador de Plan2Explore, premiado donde los ocho modelos no se ponen de acuerdo")
print("APRENDIZAJE: modelo, actor y crítico activos; evaluación con el actor aprendido")
"""
    revision = subprocess.run(
        [str(Path(REPO) / ".venv/bin/python"), "-c", comprobacion, json.dumps(list(argumentos)), TAREA],
        cwd=REPO, text=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
    )
    print(revision.stdout, end="")
    if revision.returncode != 0:
        raise RuntimeError("Configuración rechazada. El entrenamiento NO se ha iniciado.")


print("QUIÉN JUNTA LOS DATOS: EL EXPLORADOR DE PLAN2EXPLORE")
print("model.plan2explore=True | trainer.random_policy=False | model.train_actor_critic=True")
print(LOGDIR)
print(OPCIONES)
if (Path(REPO) / ".venv/bin/python").exists():
    verificar_explorador(shlex.split(OPCIONES) + [f"env.steps={PASOS}", f"logdir={LOGDIR}"])
else:
    print("Antes de entrenar se verificará esta configuración con el Python instalado en la sección 4.")
```

## 10. Entrenar (o seguir)

Arriba aparece una barra de progreso con el tiempo que falta, y debajo va corriendo el log. Al retomar, cargar el buffer desde Drive tarda unos minutos antes de que la barra aparezca.

**Para detenerlo:** usa el botón de detener de esta celda. El entrenamiento termina el paso en que va, guarda el modelo y el buffer en Drive (puede tardar varios minutos) y avisa con `Listo, quedó guardado`. No cierres la pestaña antes de ese mensaje.

**Para seguir otro día:** corre las celdas 1 a 6 y después esta. Sigue exactamente desde el paso en que lo detuviste.

**Si la sesión se corta sola** (desconexión, límite de Colab): sigue igual, desde el último guardado. El modelo se guarda cada 10.000 pasos y el buffer cada 100.000; el log avisa qué tramo de datos se perdió, y se vuelve a juntar.

Para dejarlo corriendo de noche: deja la pestaña abierta y que el PC no entre en suspensión (la pantalla sí se puede apagar). Cada guardado reemplaza el buffer anterior, que queda en la papelera de Drive; puedes vaciarla al final.

```python
import re
import subprocess

from tqdm.auto import tqdm

comando = [
    str(Path(REPO) / ".venv/bin/python"), "-u", "train.py", *shlex.split(OPCIONES),
    f"env.steps={PASOS}", f"env.eval_episode_num={PARTIDAS_EVAL}",
    f"trainer.eval_every={EVALUAR_CADA}", "trainer.update_log_every=2500", "trainer.save_every=10000",
    f"trainer.save_buffer_every={GUARDAR_BUFFER_CADA}", f"logdir={LOGDIR}",
]
if PARAR_EN_PASO is not None:
    comando.append(f"trainer.stop_at={PARAR_EN_PASO}")

# Se verifica el comando final antes de crear el proceso de entrenamiento.
verificar_explorador(comando[3:])
proceso = subprocess.Popen(comando, cwd=REPO, text=True, bufsize=1,
                           stdout=subprocess.PIPE, stderr=subprocess.STDOUT)
barra = None
detenido = False


def mostrar(linea):
    global barra
    paso = re.match(r"\[(\d+)\]", linea)
    if paso:
        paso = min(int(paso.group(1)), PASOS)
        if barra is None:
            # Parte desde el paso en que va, por si el run se retomó a mitad de camino.
            barra = tqdm(total=PASOS, initial=paso, unit="paso", smoothing=0, desc=CASO)
        barra.update(paso - barra.n)
        if "eval_score" in linea or "Saved the replay buffer" in linea:
            barra.set_postfix_str("")
        if " progress " in linea:
            return  # la barra ya muestra el avance
    if barra is not None and linea.startswith("Evaluating"):
        barra.set_postfix_str("evaluando")
    if barra is not None and "Saving the replay buffer" in linea:
        barra.set_postfix_str("guardando el buffer")
    print(linea, end="")


try:
    for linea in proceso.stdout:
        mostrar(linea)
except KeyboardInterrupt:
    detenido = True
    # train.py recibe la señal, termina el paso en que va y guarda todo antes de salir.
    proceso.terminate()
    print("\nDeteniendo: se está guardando el modelo y el buffer en Drive. Puede tardar varios minutos; "
          "no cierres la pestaña hasta que diga que quedó guardado.")
    while proceso.poll() is None:
        try:
            for linea in proceso.stdout:
                print(linea, end="")
        except KeyboardInterrupt:
            print("Ya se está guardando; espera a que termine.")
finally:
    if barra is not None:
        barra.close()
proceso.wait()
if proceso.returncode != 0:
    print(f"El entrenamiento terminó con un error (código {proceso.returncode}). Revisa las últimas líneas de arriba.")
elif detenido:
    print("Listo, quedó guardado. Para seguir otro día: celdas 1 a 6 y después esta.")

if APAGAR_AL_TERMINAR and not detenido and proceso.returncode == 0:
    from google.colab import drive, runtime

    # Drive sube los archivos con atraso: esto espera a que todo quede guardado antes de apagar.
    print("Guardando todo en Drive y apagando la sesión para que deje de gastar unidades...")
    drive.flush_and_unmount()
    runtime.unassign()
```

## 11. Explorador vs. random

Compara este run con el random de la misma vista (`LOGDIR_RANDOM`, semana del 22), hasta el paso en que va este.

- **Izquierda:** galletas que vio quien juntaba los datos, acumuladas. Es lo primero que se nota: si el explorador llega más a las galletas, el buffer tiene más ejemplos de los que aprender.
- **Derecha:** `episode/eval_score`, las galletas del actor aprendido en cada evaluación. No se suavizan ni se inventan puntos.

Si un run se retomó desde un checkpoint anterior, se usa solo el tramo más reciente.

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
from IPython.display import Image, display

sys.path.insert(0, REPO)
from compare_runs import load_run

corridas = {"Explorador (esta semana)": LOGDIR, "Random (semana del 22)": LOGDIR_RANDOM}
runs = {}
for nombre, carpeta in corridas.items():
    if (Path(carpeta) / "metrics.jsonl").exists():
        runs[nombre] = load_run(carpeta)
    else:
        print(f"{nombre}: todavía no hay métricas en {carpeta}")

if "Explorador (esta semana)" in runs:
    explorador = runs["Explorador (esta semana)"]
    hasta = max(row["step"] for row in explorador.rows)
    print(f"Hasta el paso {hasta:,}:")
    fig, (izq, der) = plt.subplots(1, 2, figsize=(13, 4.5))
    for nombre, run in runs.items():
        partidas = [(paso, galletas) for paso, galletas in run.points("episode/score") if paso <= hasta]
        apretones = [n for paso, n in run.points("episode/presses") if paso <= hasta]
        total, acumulado = 0, []
        for paso, galletas in partidas:
            total += galletas
            acumulado.append((paso, total))
        por_partida = sum(apretones) / len(apretones) if apretones else 0
        print(f"  {nombre}: vio {total:,.0f} galletas en {len(partidas):,} partidas "
              f"y apretó el botón {por_partida:.1f} veces por partida")
        if acumulado:
            izq.step(*zip(*acumulado), where="post", label=nombre)
        evaluaciones = [(paso, g) for paso, g in run.points("episode/eval_score") if paso <= hasta]
        if evaluaciones:
            der.plot(*zip(*evaluaciones), marker="o", label=nombre)
    vista = "mapa completo" if VISTA is None else f"vista {VISTA}×{VISTA}"
    izq.set_title(f"Galletas que vio quien juntaba los datos · {vista}")
    izq.set_ylabel("Galletas acumuladas")
    der.set_title(f"Evaluación del actor aprendido · {vista}")
    der.set_ylabel("Galletas por partida (media de la evaluación)")
    for ax in (izq, der):
        ax.set_xlabel("Pasos en el ambiente")
        ax.xaxis.set_major_formatter(FuncFormatter(lambda x, _: f"{x / 1000:g}k"))
        ax.set_ylim(bottom=0)
        ax.grid(alpha=0.3)
        ax.legend()
    fig.tight_layout()
    salida = Path(LOGDIR) / "explorador_vs_random.png"
    fig.savefig(salida, dpi=150)
    display(Image(str(salida)))
    plt.close(fig)
```
